# 🔢 log₁₀ (Log base 10) Rule কী?
log₁₀(N) মানে হলো— 10 কে কত power দিলে N পাওয়া যায়।

10^x = N  →  x = log₁₀(N)

Digit Count Trick (Programming)

    কোনো সংখ্যা কত digit আছে:
    digits = floor(log10(n)) + 1
    Example

    n = 5873
    log10(5873) ≈ 3.76
    floor(3.76)=3
    digits = 3+1 = 4 ✅

```python
import math

n = 5873
digits = math.floor(math.log10(n)) + 1
print(digits)
```

### What is a Palindrome Number?
A palindrome number is a number that is the same when read forward and backward.

Examples

121 → 121 ✅ Palindrome

1331 → 1331 ✅ Palindrome

123 → 321 ❌ Not Palindrome
### 🎯 Goal of This Code

Reverse the number

Compare the reversed number with the original number

If both are same → Palindrome

```python
n = 1234
num = n
result = 0

while num > 0:
    ld = num % 10
    result = (result * 10) + ld
    num = num // 10

print(n == result)

```

```python
def is_palindrome(n):
    num = n
    result = 0

    while num > 0:
        ld = num % 10
        result = (result * 10) + ld
        num = num // 10

    return n == result

print(is_palindrome(1234))
print(is_palindrome(121)) 



```

### What is an Armstrong Number?
A number is an Armstrong number if the sum of its digits each raised to the power of the number of digits equals the number itself.

Example:




```python
num = 153
digits = len(str(num))
sum_of_powers = 0
temp = num

while temp > 0:
    digit = temp % 10
    sum_of_powers += digit ** digits
    temp //= 10

if sum_of_powers == num:
    print(num, "is an Armstrong number")
else:
    print(num, "is not an Armstrong number")

```
