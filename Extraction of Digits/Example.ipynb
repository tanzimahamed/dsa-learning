# Count Digits (Digit কতগুলো আছে)
Logic:

👉 যতবার loop চলবে = digit সংখ্যা

```python
n = 5873
count = 0

while n > 0:
    n = n // 10
    count += 1

print(count)
```

# 🔢 log₁₀ (Log base 10) Rule কী?
log₁₀(N) মানে হলো— 10 কে কত power দিলে N পাওয়া যায়।

10^x = N  →  x = log₁₀(N)

Digit Count Trick (Programming)

    কোনো সংখ্যা কত digit আছে:
    digits = floor(log10(n)) + 1
    Example

    n = 5873
    log10(5873) ≈ 3.76
    floor(3.76)=3
    digits = 3+1 = 4 ✅

```python
import math

n = 5873
digits = math.floor(math.log10(n)) + 1
print(digits)
```
